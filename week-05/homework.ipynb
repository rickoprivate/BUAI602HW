**Task:**

Write a function `total_length(words)` that returns the sum of the lengths of all words in a list using the accumulator pattern.

Write a second version using list comprehension.

**Sample Output:**

```

```
should return `15`




`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def total_length(words):
    '''calculate the sum of the lengths of all words in the provided list format'''
    sum_lengths = 0
    for word in words:
        '''accumulator for all word lengths'''
        sum_lengths += len(word)
    return sum_lengths
`````

**Review Problem 3**

**Concept:** *List processing, accumulation, decision statements, creating a new list as result.*

**Task:**

Write a function `union(a, b)` that returns a new list containing all unique elements from both lists.

Rules:

* Cannot use a set

* Cannot modify input lists

* Must create a new list

* Order does not matter

**Sample Output:**

```
union of [1] with [1,2] should return [1, 2]
union of [1, 3] with [2, 4, 5] should return [1, 2, 3, 4, 5]
union of [4, 5] [1, 2, 3, 4] should return [4, 5, 1, 2, 3]
```




`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def union(a, b):
    '''use union of two lists'''
    new_list = []
    for x in a, b:
        '''loop loop through given lists'''
        for x2 in x:
            '''loop throug elements in each list'''
            if x2 not in new_list:
                new_list.extend([x2])

    return new_list
`````

`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def union_2(a, b):
    '''return unique elements by a list, convert into a set then back to list'''
    rem_keys = []
    ''' create a list of unique keys'''
    rem_keys = set(a) | set(b)
    rem_keys = list(rem_keys)

    return rem_keys
`````

`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def word_score(word):
    ''' return the scrabble score for the given string, calls on scrable_scores function'''
    score = 0
    for chr in word:
        '''accumulate the returned scores'''
        score += scrabble_scores[chr]
    return score
`````

**Problem 9**

**Concept:** *Using list data type, dictionary data type, loop, dictionary element access, accumulator pattern.*

**Task:**

Write a function `total_points(games_results, team_name)` that returns the total points scored by a team across all games.

**Sample Output:**
```
games_results = [
    {'A': 5, 'B': 7},
    {'A': 10, 'C': 3}
]
print(total_points(games_results, 'A'))
```
should return
```
total points for A: 15
```



`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def total_points(game_results, team_name):
    '''return the accumlated total score for a given team names from the given dictionary'''
    scores_list = [s[team_name] for s in game_results]
    total_score = 0
    for score in scores_list:
        '''accumulate the scores'''
        total_score += score

    return total_score
`````

`````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def sort_games_by_score(games, team_name):
    '''return a descending list of sorted scores for a given name and given dictionary'''
    sorted_scores = sorted(games, key=lambda s: s[team_name], reverse=True)
    return sorted_scores
`````
